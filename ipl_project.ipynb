```python
matches = pd.concat([ipl['team1'],ipl['team2']]).value_counts()
```

```python
matches
```

```python
wins = ipl['winner'].value_counts()
```

```python
wins
```

```python
losses = matches - wins
```

```python
losses
```

```python
team_stats = pd.DataFrame({
    'matches' : matches,
    'wins' : wins,
    'losses' : losses
})
```

```python
team_stats
```

```python
team_stats.sort_values(['matches','losses'],ascending=False)
```

```python
ax = team_stats[['wins', 'losses']].plot(
    kind='bar',
    figsize=(12, 6),
    title='IPL Teams - Wins vs Losses'
)

ax.set_xlabel("Teams")
ax.set_ylabel("Number of Matches")

# Add values on bars
for container in ax.containers:
    ax.bar_label(container, padding=5)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
